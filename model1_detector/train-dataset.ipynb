**valid/image folder doesn't exist so lets fix that shouldn't we**

```python
import os, shutil

working_dir = '/kaggle/working/merged_dataset'

# Wipe the folder if you are re-running the notebook to avoid errors
if os.path.exists(working_dir):
    shutil.rmtree(working_dir)

print("Copying dataset to working directory so YOLO can write cache files...")

# IMPORTANT: Make sure this path matches exactly what you named your new Kaggle dataset!
shutil.copytree(
    '/kaggle/input/datasets/malaklotfy/merged-dataset/merged_dataset',
    working_dir
)

print("✅ Dataset copied successfully! Ready for YOLO.")
```

**TRAIN AND SAVE LOCALLY**

```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt')

results = model.train(
    data='/kaggle/working/merged_dataset/dataset.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    device='cuda',
    project='/kaggle/working/recipe_runs',
    name='ingredient_detector',
    exist_ok=True,
)

print('DONE. mAP50:', results.results_dict['metrics/mAP50(B)'])
```
